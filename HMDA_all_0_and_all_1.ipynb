```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("C:/Users/gdani/OneDrive/Desktop/master thesis/datasets/HDMA/X_df_selected.csv", delimiter=",")
df
```

```python
df['applicant_sex'] = df['applicant_sex'].apply(lambda x: 1 if x in [1, 3] else 2)
print(df['applicant_sex'].unique())

unique_count = df['income'].nunique()
print(f"Number of unique values of income: {unique_count}")

unique_values=df['income'].unique()
mean_unique = unique_values.mean()
print(f"Mean of all unique income values: {mean_unique}")


mean = df['income'].mean()
print(f"Mean income: {mean}")
```

```python
# Identify non-numeric columns
non_numeric = df.select_dtypes(exclude=['number'])

# Count them
num_non_numeric = non_numeric.shape[1]

print(f"Number of non-numeric variables: {num_non_numeric}")
```

```python
#Feature selection

#Convert non numeric values in numeric values
df= df.apply(pd.to_numeric, errors='coerce')

#Check datatypes
print("\n",df.info())
```

```python
# Identify non-numeric columns
non_numeric = df.select_dtypes(exclude=['number'])

# Count them
num_non_numeric = non_numeric.shape[1]

print(f"Number of non-numeric variables: {num_non_numeric}")
```

```python
#drop missing values
df = df.dropna()
#Check
print("\n",df.info())
```

```python
# Function that always returns 0 (negative prediction)
def all_no_model(row):
    return 0

# Apply the model to create predictions
df['prediction'] = df.apply(all_no_model, axis=1)

# Check unique values in prediction column
print(df['prediction'].unique())  # Should print: [0]
```

```python
# Function that always returns 1 (positive prediction)
def all_no_model(row):
    return 1

# Apply the model to create predictions
df['prediction'] = df.apply(all_no_model, axis=1)

# Check unique values in prediction column
print(df['prediction'].unique())  # Should print: [1]
```
