```python
#drop missing values
df = df.dropna()
#Check
print("\n",df.info())
```

```python
# Function to create a rule-based biased model with respect to equalized odds
def biased_model(row):
    # Set decision rules
    if row["applicant_sex"] == 1:  # Lenient threshold for males
        if row['income'] >=0.1521273322710113:         # Remember: the average income is 1.8448083668323002e-17, it's super small
            return 1                                   # We'll use half of the mean of all'unique values as a trashold
        else:
            return 0
    else:  # Stricter threshold for females
        if row['income'] >= 1.363403731633618:
            return 1
        else:
            return 0

# Apply the biased model to the DataFrame to create predictions
df['prediction'] = df.apply(biased_model, axis=1)
```

```python
# Rule to dalcolate true and false positive rates
# Remember that respecting equalised odds means having the same TPR and FPR among both protected groups
def calculate_tpr_fpr(group):
    tp = ((group['prediction'] == 1) & (group['target'] == 1)).sum()
    fn = ((group['prediction'] == 0) & (group['target'] == 1)).sum()
    fp = ((group['prediction'] == 1) & (group['target'] == 0)).sum()
    tn = ((group['prediction'] == 0) & (group['target'] == 0)).sum()

    try:
        tpr = tp / (tp + fn)
    except ZeroDivisionError:
        tpr = 0  # If there's a division by zero, return 0

    try:
        fpr = fp / (fp + tn)
    except ZeroDivisionError:
        fpr = 0  # If there's a division by zero, return 0

    return tpr, fpr

# Group by 'applicant_sex' and calculate TPR/FPR
for applicant_sex_value in df['applicant_sex'].unique():
    group = df[df['applicant_sex'] == applicant_sex_value]
    tpr, fpr = calculate_tpr_fpr(group)

    # Print the TPR and FPR for each group
    print(f"Group {applicant_sex_value} — TPR: {tpr:.2f}, FPR: {fpr:.2f}")
```
